## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

###Answer
**Decision Improved:**
This work improves the decision of which declining web pages should be prioritized for a content review and refresh during the next planning cycle.

**Who Acts:**
SEO managers and content editors act on these predictions. They update outdated information, rewrite meta titles and descriptions, and fix search intent gaps on the flagged pages.

**Cost of a Wrong Call:**
A wrong recommendation carries costs on both sides. A **false positive** wastes writer time and budget editing a healthy page that was already performing well. A **false negative** leaves high-visibility pages unnoticed, causing a quiet loss of organic search traffic and conversion revenue.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import sys
import subprocess
import pandas as pd

# Setup repository environment and working directory
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

# Load the dataset from the root directory
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Calculate metrics
total_pages = len(df)
declining_pages = (df['trend_direction'] == 'down').sum()
declining_pct = (declining_pages / total_pages) * 100
high_demand_declining = len(df[(df['impressions_90d'] >= 500) & (df['trend_direction'] == 'down')])

print(f"Total pages: {total_pages:,}")
print(f"Declining pages: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"High-demand declining pages (>=500 impressions): {high_demand_declining:,}")
```

In the 30,000 total pages evaluated in the starter dataset, 16,262 pages (54.2%) show an active downward trend.

Out of those declining pages, 9,961 still receive high exposure with at least 500 impressions over 90 days.

These numbers show that over half of the catalog—and nearly 10,000 high-visibility pages—are actively losing momentum, making a machine learning prioritization queue well worth building over the next 7 weeks.
